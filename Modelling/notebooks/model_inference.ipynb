```python
import pickle
import pandas as pd

# Load Trained Pipeline Artifact
with open('artifacts/best_churn_pipeline.pkl', 'rb') as f:
    pipeline = pickle.load(f)

print("TEST MODEL PIPELINE ARTIFACT")

# INPUT DATA FOR INFERENCE
new_customer = pd.DataFrame([{
    'CreditScore': 640,
    'Geography': 'France',
    'Gender': 'Female',
    'Age': 41,
    'Tenure': 4,
    'Balance': 75000.0,
    'NumOfProducts': 1,
    'HasCrCard': 1,
    'IsActiveMember': 0,
    'EstimatedSalary': 50000.0,
    'Balance_Salary_Ratio': 1.5,
    'USD_TO_IDR': 15800.0
}])

# Predict via Pipeline
pred_class = pipeline.predict(new_customer)
pred_proba = pipeline.predict_proba(new_customer)[:, 1]

print("\n=== RESULT OF MODEL INFERENCE ===")
print(f"Prediction churn status: {'CHURN (1)' if pred_class[0] == 1 else 'RETAINED (0)'}")
print(f"Churn Probability: {pred_proba[0] * 100:.2f}%")
```
